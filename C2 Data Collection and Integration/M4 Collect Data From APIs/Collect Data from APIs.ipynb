```python
import yfinance as yf
google_info = yf.Ticker('GOOGL')
```

```python
print(google_info.info)
```

```python
print(f'Company sector: {google_info.info['sector']}')
```

```python
for key, value in google_info.info.items():
    print(f'{key}: {value}')
```

```python
print(google_info.history(period = 'max'))
```

```python
import datetime
start = datetime.datetime(2012, 5, 31)
end = datetime.datetime(2013, 1, 30)
print(google_info.history(start = start, end = end))
```

```python
df = yf.download('AMZN GOOGL', start = '2019-01-01', end = '2019-01-03', group_by = 'ticker')
print(df)
print(df.GOOGL)
```

```python
df.to_csv('Data/AMZNGOOGL.csv')
```
