### Web Scraping

```python
import requests
```

```python
page = requests.get('https://dataquestio.github.io/web-scraping-pages/simple.html')
page
```

```python
page.content
```

```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(page.content, 'html.parser')
```

```python
print(soup.prettify())
```

```python
soup.find_all('title')
```

```python
for text in soup.find_all('p'):
    print(text.get_text())
```

```python
soup.find('p').get_text()
```
