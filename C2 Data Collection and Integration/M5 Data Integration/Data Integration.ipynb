### Concat

```python
df1 = df[['longitude', 'latitude', 'median_income']].sample(n = 5)
df2 = df[['longitude', 'latitude', 'median_income']].sample(n = 5)
df3 = df[['latitude', 'longitude', 'median_income']].sample(n = 5)
```

```python
df_concat1 = pd.concat([df1, df2, df3], axis=0)
```

```python
df_concat1
```

```python
df_concat1.info()
```

```python
df_concat2 = pd.concat([df1, df2, df3], axis = 1)
```

```python
df_concat2
```

```python
df_concat2.info()
```

```python
df1 = df[['longitude']][:5]
df2 = df[['latitude']][:5]
df3 = df[['median_income']][:5]
```

```python
df_concat1 = pd.concat([df1, df2, df3], axis = 1)
df_concat1
```

### Merge

```python
df1 = df[['longitude','median_income']][:5]
df1
```

```python
df2 = df[['longitude','median_house_value']][:5]
df2
```

```python
df_merge1 = pd.merge(df1, df2, on = ['longitude'], how = 'inner')
df_merge1
```

```python
df3 = df[['longitude','population']][2:7]
df3
```

```python
df_merge2 = pd.merge(df_merge1, df3, on=['longitude'], how='inner')
df_merge2
```

```python
df_merge3 = pd.merge(df_merge1, df3, on=['longitude'], how='left')
df_merge3
```

```python
df_merge4 = pd.merge(df_merge1, df3, on=['longitude'], how='outer')
df_merge4
```
